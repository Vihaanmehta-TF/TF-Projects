```python
import pandas as pd
import matplotlib.pyplot as plt
survey = pd.read_csv("class_survey.csv")
counts = survey["favorite_subject"].value_counts()
plt.bar(counts.index, counts.values, color="#2A9D8F")
plt.title("Students' favorite subject")
plt.xlabel("Subjects")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.show()
print("History and PE are tied for the Students' favorite Subject with 8 votes each, and Music has the least votes with 5.")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
plt.hist(survey["minutes_reading"], bins = 8, color="#F26419", edgecolor ="white")
plt.title("Students' time reading (in minutes)")
plt.xlabel("Number of Minutes Reading")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.show()
print("The time range with the most students in it was tied between 10-19, 20-29, and 50-59")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
plt.scatter(survey["hours_sleep"], survey["minutes_gaming"], color="#C1146B")
plt.title("Students' minutes gaming and hours of sleep")
plt.xlabel("Hours of sleep")
plt.ylabel("Minutes gaming")
plt.tight_layout()
plt.show()
print("This scatter plot just looks completely random.")
```
